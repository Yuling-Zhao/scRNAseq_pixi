```r
library(ggplot2)
```

```r
x <- 1:100
y <- cumsum(rnorm(100))

plot(
  x, y,
  type = "l",
  lwd = 2,
  main = "R Notebook Test",
  xlab = "Index",
  ylab = "Value"
)
```

```r
library(ggplot2)

df <- data.frame(
  x = 1:100,
  y = cumsum(rnorm(100))
)

ggplot(df, aes(x, y)) +
  geom_line() +
  ggtitle("ggplot2 Test")
```
